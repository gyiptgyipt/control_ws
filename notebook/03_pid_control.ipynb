# PID Control: Interactive Learning

**Plant:** G(s) = 1 / (s(s+2)(s+5))  (Type 1 — integrator in plant)
**Controller:** C(s) = Kp + Ki/s + Kd·s

**Drag sliders to tune PID and see closed-loop response.**

---

```python
# Run this cell first to set up the interactive plot
import sys
sys.path.insert(0, '/home/zyme/control_ws/MathPlot')

import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
import ipywidgets as widgets
from IPython.display import display

%matplotlib widget

def pid_closed_loop(Kp, Ki, Kd, sim_time=10):
    plant_num = [1]
    plant_den = [1, 7, 10, 0]  # s(s+2)(s+5) = s³ + 7s² + 10s
    ctrl_num = [Kd, Kp, Ki]
    ctrl_den = [1, 0]  # s
    ol_num = np.polymul(ctrl_num, plant_num)
    ol_den = np.polymul(ctrl_den, plant_den)
    cl_num = ol_num
    cl_den = np.polyadd(ol_den, ol_num)
    sys_cl = signal.TransferFunction(cl_num, cl_den)
    t = np.linspace(0, sim_time, 1000)
    t, y = signal.step(sys_cl, T=t)
    return t, y, sys_cl

def plot_pid(Kp=15, Ki=8, Kd=2):
    t, y, sys_cl = pid_closed_loop(Kp, Ki, Kd)
    
    fig, axes = plt.subplots(2, 1, figsize=(10, 8))
    
    # Output response
    ax1 = axes[0]
    ax1.plot(t, y, 'b-', linewidth=2, label='Output')
    ax1.axhline(y=1, color='k', linestyle='--', linewidth=1, label='Reference')
    
    ss = y[-1]
    overshoot = max(y) - ss
    po = (overshoot / ss) * 100 if ss > 0 else 0
    
    idx_10 = np.where(y >= 0.1*ss)[0]
    idx_90 = np.where(y >= 0.9*ss)[0]
    rise_time = t[idx_90[0]] - t[idx_10[0]] if len(idx_10) > 0 and len(idx_90) > 0 else np.nan
    
    settling_idx = np.where(np.abs(y - ss) <= 0.02*ss)[0]
    settling_time = t[settling_idx[-1]] if len(settling_idx) > 0 else np.nan
    
    title_str = f'PID Closed-Loop: Kp={Kp:.1f}, Ki={Ki:.1f}, Kd={Kd:.1f}'
    title_str += f'\nPO={po:.1f}%, tr={rise_time:.2f}s, ts={settling_time:.2f}s, ess={1-ss:.4f}'
    ax1.set_ylabel('Output', fontsize=12)
    ax1.set_title(title_str, fontsize=12)
    ax1.grid(True, alpha=0.3)
    ax1.legend(loc='lower right')
    ax1.set_xlim(0, 10)
    ax1.set_ylim(-0.2, max(1.5, max(y)*1.1))
    
    # Control effort
    ax2 = axes[1]
    ctrl_num = [Kd, Kp, Ki]
    ctrl_den = [1, 0]
    plant_num = [1]
    plant_den = [1, 7, 10, 0]
    ol_num = np.polymul(ctrl_num, plant_num)
    ol_den = np.polymul(ctrl_den, plant_den)
    sens_num = ol_den
    sens_den = np.polyadd(ol_den, ol_num)
    err_num = sens_num
    err_den = np.polymul(sens_den, [1, 0])
    u_num = np.polymul(ctrl_num, err_num)
    u_den = np.polymul(ctrl_den, err_den)
    sys_u = signal.TransferFunction(u_num, u_den)
    t, u = signal.step(sys_u, T=t)
    
    ax2.plot(t, u, 'r-', linewidth=2, label='Control Effort')
    ax2.set_xlabel('Time (s)', fontsize=12)
    ax2.set_ylabel('Control Signal', fontsize=12)
    ax2.set_title(f'Max Control Effort: {max(np.abs(u)):.2f}', fontsize=12)
    ax2.grid(True, alpha=0.3)
    ax2.legend()
    ax2.set_xlim(0, 10)
    
    plt.tight_layout()
    plt.show()
    
    print(f'\nPID Performance:')
    print(f'  Kp={Kp:.1f}, Ki={Ki:.1f}, Kd={Kd:.1f}')
    print(f'  Steady-state: {ss:.4f}')
    print(f'  Overshoot: {po:.1f}%')
    print(f'  Rise time (10-90%): {rise_time:.2f}s')
    print(f'  Settling time (2%): {settling_time:.2f}s')
    print(f'  Steady-state error: {1-ss:.4f}')
    print(f'  Max control effort: {max(np.abs(u)):.2f}')

# Create interactive sliders
Kp_slider = widgets.FloatSlider(value=15, min=0, max=50, step=1,
    description='Kp:', style={'description_width': 'initial'}, continuous_update=True)
Ki_slider = widgets.FloatSlider(value=8, min=0, max=50, step=1,
    description='Ki:', style={'description_width': 'initial'}, continuous_update=True)
Kd_slider = widgets.FloatSlider(value=2, min=0, max=10, step=0.5,
    description='Kd:', style={'description_width': 'initial'}, continuous_update=True)

ui = widgets.VBox([
    widgets.HTML('<h3>PID Control: Plant G(s) = 1/(s(s+2)(s+5))</h3>'),
    widgets.HTML('<i>C(s) = Kp + Ki/s + Kd·s  |  Tune for low overshoot, fast settling, zero steady-state error</i>'),
    Kp_slider, Ki_slider, Kd_slider
])

out = widgets.interactive_output(plot_pid, {'Kp': Kp_slider, 'Ki': Ki_slider, 'Kd': Kd_slider})
display(ui, out)
```

## PID Term Effects

| Term | Effect | Trade-off |
|------|--------|-----------|
| **Kp** (Proportional) | Reduces rise time, reduces steady-state error | Increases overshoot, may cause instability |
| **Ki** (Integral) | **Eliminates steady-state error** | Increases overshoot, slows response, oscillation risk |
| **Kd** (Derivative) | Reduces overshoot, improves stability | Amplifies noise, sensitive to measurement noise |

## Manual Tuning Steps

1. **Start**: Kp=1, Ki=0, Kd=0
2. **Increase Kp** until sustained oscillation, then halve it
3. **Add Ki** to eliminate steady-state error (increase until oscillation, then reduce)
4. **Add Kd** to dampen overshoot (small values help)

## Good Starting Point for This Plant

- **Kp = 15, Ki = 8, Kd = 2** — reasonable balance
- Try **Kp = 25, Ki = 15, Kd = 3** — more aggressive
- Try **Kp = 10, Ki = 5, Kd = 1** — more conservative

## Try These Experiments

1. **Kp only** (Ki=0, Kd=0): Increase Kp — watch steady-state error decrease but overshoot increase
2. **Add Ki**: Set Kp=10, increase Ki from 0 — steady-state error goes to zero, but overshoot grows
3. **Add Kd**: With Kp=15, Ki=8, increase Kd — overshoot decreases
4. **Too much Ki**: Kp=15, Ki=30, Kd=2 — oscillation/windup
5. **Target specs**: Try to achieve PO<10%, ts<3s, ess≈0
